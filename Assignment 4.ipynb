Q1 Stop sending emails after 3 users (Using for loop and break)

users = ["Ali", "Sara", "John", "Zara", "Mike"]

Output:

Email sent to Ali

Email sent to Sara

Email sent to John

Limit reached. Stop sending emails.

```python
users = ["Ali", "Sara", "John", "Zara", "Mike"]
num = 0
for i in users:
    if num == 3:
        print("Limit reached. Stop sending emails")
        break
    else:
        print(f" Email sent to {i}")
        num +=1
        
```

Q2. Stop monitoring temperature if critical temperature reached (Using for loop and break)

temps = [22, 25, 30, 45, 28] critical = 40

Ouput:

Temperature: 22

Temperature: 25

Temperature: 30

Critical temperature reached! Stop monitoring.

```python
temps = [22, 25, 30, 45, 28] 
for i in temps:
    if i >= 40:
        print("Critical Temperature reached stop monitoring")
        break
    else:
        print(f'Temperature : {i}')
        
```

Q4. Skip customers with unpaid bills (Using For loop and continue)

customers = [ {"name": "Ali", "paid": True},

{"name": "Sara", "paid": False},

{"name": "John", "paid": True}
]

Output:

Access granted to: Ali

Access granted to: John

```python
customers = [ {"name": "Ali", "paid": "True"},

{"name": "Sara", "paid": "False"},

{"name": "John", "paid": "True"} ]
for i in customers:
    if i['paid'] == "False":
        continue
    else:
        print(f"Access granted to :{i['name']}")
        
```

Q6. Extract domain names from email list (Using for loop)

emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]

Output:

domains: ['gmail.com', 'yahoo.com', 'hotmail.com']



```python
emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]
domain = []
for i in emails:
    domains=i.split("@")[1]
    domain.append(domains)
print(domain)
```

Q8. Create a multiplication table using a nested for loop.

numbers = [2, 3, 4]
multipliers = [1, 2, 3]

Expected Output:

2 x 1 = 2
2 x 2 = 4
2 x 3 = 6
3 x 1 = 3
3 x 2 = 6
3 x 3 = 9
4 x 1 = 4
4 x 2 = 8
4 x 3 = 12

Use a nested for loop.

```python
numbers = [2, 3, 4] 
multipliers = [1, 2, 3]
for  i in numbers:
    for x in multipliers:
        print(f"{i} X {x} = ", i*x)
```

Q9. Keep asking the user for a password until the correct password is entered.

correct_password = "python123"

Expected behaviour:

Enter password: abc
Incorrect password. Try again.

Enter password: 12345
Incorrect password. Try again.

Enter password: python123
Access granted!

Use a while loop.

```python
correct_password = "python123"
password = ""
while password != correct_password:
    password = input("Enter password: ")
    if password != correct_password:
        print("Incorrect password. Try again.")
    else:
        print("Access granted!")
```

Q10. Count down from 5 to 1 and then display "Time's up!".

Expected Output:

5
4
3
2
1
Time's up!

Use a while loop.

```python
count = 5

while count >= 1:
    print(count, end=" ")
    count -= 1

print("Time's up!")
```
