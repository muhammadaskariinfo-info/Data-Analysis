```python
products = {
    "P1": {"name": "Shoes", "price": 2500, "stock": 5},
    "P2": {"name": "Shirt", "price": 1500, "stock": 10},
    "P3": {"name": "Bag", "price": 3000, "stock": 4},
    "P4": {"name": "Watch", "price": 4000, "stock": 3},
    "P5": {"name": "Headphones", "price": 2000, "stock": 6}
}

cart = {}

while True:

    print("\n1. View Products")
    print("2. Add to Cart")
    print("3. View Cart")
    print("4. Checkout")
    print("5. Exit")

    choice = input("Enter choice: ")

    if choice == "1":

        for id in products:
            print(id, products[id]["name"],
                  products[id]["price"],
                  products[id]["stock"])

    elif choice == "2":

        id = input("Enter Product ID: ")
        qty = int(input("Enter Quantity: "))

        if id in products:

            if qty > 0 and qty + cart.get(id, 0) <= products[id]["stock"]:

                cart[id] = cart.get(id, 0) + qty
                print("Added to Cart!")

            else:
                print("Not enough stock!")

        else:
            print("Product not found!")

    elif choice == "3":

        total = 0

        for id in cart:

            qty = cart[id]
            price = products[id]["price"]
            subtotal = qty * price

            print(products[id]["name"], qty, subtotal)

            total = total + subtotal

        print("Cart Total:", total)

    elif choice == "4":

        total = 0

        for id in cart:
            total += products[id]["price"] * cart[id]

        if total == 0:
            print("Cart is empty!")

        else:

            if total >= 5000:
                discount = total * 0.10
            else:
                discount = 0

            final = total - discount

            print("Total:", total)
            print("Discount:", discount)
            print("Final Bill:", final)

            for id in cart:
                products[id]["stock"] -= cart[id]

            cart.clear()

            print("Checkout Successful!")

    elif choice == "5":
        print("Thank you!")
        break

    else:
        print("Invalid Choice!")

 
```
