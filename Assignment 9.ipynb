## ***16. Find Average BasePay Of All Employee Per Year***

```python
df.info()
```

```python
df.BasePay=df.BasePay.replace("Not provided" , 'NaN')
```

```python
df["BasePay"] = pd.to_numeric(df["BasePay"], errors="coerce")
avg = df["BasePay"].mean()
print(avg) 
```

## ***17. Find Average BasePay Of All Employee Per Job Title***

```python
avg_basepay_by_job = df.groupby("JobTitle")["BasePay"].mean()
print(avg_basepay_by_job)
```

## ***18. Find Average BasePay Of Employee Having Job Title ACCOUNTANT***

```python
avg_accountant_basepay = df.loc[df["JobTitle"] == "ACCOUNTANT", "BasePay"].mean()
print(avg_accountant_basepay)  # 46643.172
```
